```python
from transformers import pipeline
import torch
# Install TextBlob data
import nltk
nltk.download('punkt')
nltk.download('stopwords')


from textblob import TextBlob

texts = [
    "I feel really happy today, everything is going well",
    "I feel so alone and nobody understands me",
    "I am so angry, everything keeps going wrong",
    "I don't know how I feel today, just existing",
    "I am anxious about everything, I can't stop worrying"
]
emotion = pipeline(task="text-classification",model="j-hartmann/emotion-english-distilroberta-base",top_k=1)

def get_sentiment(polarity):
    if polarity > 0.2:
        return "Positive"
    elif polarity < -0.1:
        return "Negative"
    else:
        return "Neutral"
for text in texts:
    blob = TextBlob(text)
    emotions = emotion(text)
    sentiment = get_sentiment(blob.sentiment.polarity)
    print(f"Text: {text[:40]}...")
    print(f"Sentiment: {sentiment}")
    print(f"Emotion: {emotions}")
    print("---")
```
