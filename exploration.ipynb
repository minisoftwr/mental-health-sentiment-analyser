```python
# Install TextBlob data
import nltk
nltk.download('punkt')
nltk.download('stopwords')

from textblob import TextBlob

# Text Negative sentiment
texts = [
    "I feel really happy today, everything is going well",
    "I feel so alone and nobody understands me",
    "I am so angry, everything keeps going wrong",
    "I don't know how I feel today, just existing",
    "I am anxious about everything, I can't stop worrying"
]
def get_sentiment(polarity):
    if polarity > 0.2:
        return "Positive"
    elif polarity < -0.1:
        return "Negative"
    else:
        return "Neutral"
for text in texts:
    blob = TextBlob(text)
    sentiment = get_sentiment(blob.sentiment.polarity)
    print(f"Text: {text[:40]}...")
    print(f"Sentiment: {sentiment}")
    print("---")
```
